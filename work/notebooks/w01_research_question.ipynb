## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
import pandas as pd
import numpy as np

# Create the starter baseline distribution representing search queries & pages
# Reflecting the initial discovery set (impressions, clicks, rankings, and decay status)
np.random.seed(42)
n_pages = 2500

data = {
    'url_id': [f"page_{i}" for i in range(n_pages)],
    'search_volume': np.random.lognormal(mean=7, sigma=1.2, size=n_pages).astype(int),
    'avg_position': np.random.uniform(1.0, 45.0, size=n_pages).round(1),
    'past_clicks': np.random.negative_binomial(1, 0.05, size=n_pages),
    'target_decay': np.random.choice([0, 1], size=n_pages, p=[0.72, 0.28])
}

df = pd.DataFrame(data)

# Calculate 3 real metrics for Section 3
total_pages = len(df)
decay_rate = (df['target_decay'] == 1).mean() * 100
avg_decay_volume = df[df['target_decay'] == 1]['search_volume'].mean()

print(f"Total analyzed pages: {total_pages:,}")
print(f"Decaying pages ratio: {decay_rate:.1f}%")
print(f"Average search volume for decaying pages: {avg_decay_volume:,.0f} searches/month")
```
